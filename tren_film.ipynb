```python
# Import library
import requests
import pandas as pd

# Tautan Wikipedia film Indonesia tahun 2023
url = "https://id.wikipedia.org/wiki/Daftar_film_Indonesia_tahun_2023"

# Header untuk request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/131.0.0.0 Safari/537.36"
}

# Mengambil halaman Wikipedia
response = requests.get(url, headers=headers)

# Mengecek apakah halaman berhasil diakses
print(response.status_code)
```

```python
# Import library
from bs4 import BeautifulSoup
from io import StringIO

# Membaca HTML halaman
soup = BeautifulSoup(response.text, "html.parser")

# Mengambil tabel dari halaman Wikipedia
tables = pd.read_html(StringIO(str(soup)))

# Melihat jumlah tabel
print("Jumlah tabel:", len(tables))
```

```python
# Mengecek nama kolom setiap tabel tahun 2023
for i in range(len(tables)):
    print(f"\n===== TABEL {i} =====")
    print(tables[i].columns)
```

```python
# Menampilkan data awal dari setiap tabel
for i in range(len(tables)):
    print(f"\n===== TABEL {i} =====")
    display(tables[i])
```

```python
# Menggabungkan tabel film tahun 2023 (Januari-Maret, April-Juni, Juli-September, Oktober-Desember)
df_2023 = pd.concat(
    [tables[2], tables[3], tables[4], tables[5]],
    ignore_index=True
)
```

```python
# Menampilkan seluruh data film 2023 SETELAH DIGABUNGKAN
df_2023
```

```python
# Melihat nama kolom pada data film tahun 2023 (setelah digabungkan)
df_2023.columns
```

```python
# Membuat kolom baru 'Tanggal Tayang' dari tanggal, bulan, dan tahun
df_2023["Tanggal Tayang"] = (
    df_2023["Tayang.1"].astype("Int64").astype(str)
    + " "
    + df_2023["Tayang"].str.title()
    + " 2023"
)
```

```python
# Menghapus kolom 'Tayang' dan 'Tayang.1' karena sudah menjadi kolom baru 'Tanggal Tayang'
df_2023 = df_2023.drop(columns=["Tayang", "Tayang.1"])
```

```python
# Menampilkan seluruh data film tahun 2023 SETELAH menghapus kolom 'Tayang' dan 'Tayang.1'
df_2023
```

```python
# Memindahkan kolom 'Tanggal Tayang' ke posisi paling awal
kolom = ["Tanggal Tayang"] + [
    col for col in df_2023.columns
    if col != "Tanggal Tayang"
]

df_2023 = df_2023[kolom]
```

```python
# Menampilkan seluruh data film 2023 setelah dipreprocessing
df_2023
```

```python
# Memilih kolom yang diperlukan untuk dianalisis
df_2023 = df_2023[
    ["Tanggal Tayang", "Judul", "Genre", "Penonton"]
]
```

```python
# Menampilkan isi data film tahun 2023 dari kolom yang diperlukan
df_2023
```

```python
# Mengecek missing value pada setiap kolom
df_2023.isnull().sum()
```

```python
# Menghapus baris yang memiliki missing value
df_2023 = df_2023.dropna()
```

```python
# Mengecek kembali missing value
df_2023.isnull().sum()
```

```python
# Merapikan nomor index setelah data cleaning
df_2023 = df_2023.reset_index(drop=True)

# Menampilkan data film tahun 2023
df_2023
```

# SCRAPING FILM TAHUN 2024

```python
# Import library
from bs4 import BeautifulSoup
from io import StringIO

# Membaca HTML halaman
soup = BeautifulSoup(response.text, "html.parser")

# Mengambil tabel dari halaman Wikipedia
tables = pd.read_html(StringIO(str(soup)))

# Melihat jumlah tabel
print("Jumlah tabel:", len(tables))
```

```python
# Mengecek nama kolom setiap tabel tahun 2024
for i in range(len(tables)):
    print(f"\n===== TABEL {i} =====")
    print(tables[i].columns)
```

```python
# Menampilkan data awal dari setiap tabel
for i in range(len(tables)):
    print(f"\n===== TABEL {i} =====")
    display(tables[i])
```

```python
# Menampilkan seluruh data film tahun 2024 (setelah digabungkan)
df_2024
```

```python
# Melihat nama kolom pada data film tahun 2024 SETELAH DIGABUNGKAN
df_2024.columns
```

```python
# Membuat kolom baru 'Tanggal Tayang' dari tanggal, bulan, dan tahun
df_2024["Tanggal Tayang"] = (
    df_2024["Tanggal penayangan.1"].astype("Int64").astype(str)
    + " "
    + df_2024["Tanggal penayangan"].str.title()
    + " 2024"
)
```

```python
# Menghapus kolom 'Tayang' dan 'Tayang.1' karena sudah digabung menjadi kolom baru 'Tanggal Tayang'
df_2024 = df_2024.drop(columns=["Tanggal penayangan.1", "Tanggal penayangan"])
```

```python
# Menampilkan seluruh data film 2024 SETELAH menghapus kolom 'Tayang' dan 'Tayang.1'
df_2024
```

```python
# Memindahkan kolom 'Tanggal Tayang' ke posisi paling awal
kolom = ["Tanggal Tayang"] + [
    col for col in df_2024.columns
    if col != "Tanggal Tayang"
]

df_2024 = df_2024[kolom]
```

```python
# Menampilkan seluruh data film 2024 setelah preprocessing
df_2024
```

```python
# Memilih kolom yang diperlukan untuk analisis
df_2024 = df_2024[
    ["Tanggal Tayang", "Judul", "Genre", "Penonton"]
]
```

```python
# Menampilkan isi data film tahun 2024 dari kolom yang diperlukan
df_2024
```

```python
# Mengecek jumlah missing value pada setiap kolom
df_2024.isnull().sum()
```

```python
# Menghapus baris yang memiliki missing value
df_2024 = df_2024.dropna()
```

```python
# Mengecek kembali missing value
df_2024.isnull().sum()
```

```python
# Merapikan nomor index setelah data cleaning
df_2024 = df_2024.reset_index(drop=True)

# Menampilkan data film tahun 2024
df_2024
```

# Menggabungkan data film tahun 2023 & 2024

```python
# Menggabungkan data film tahun 2023 dan 2024
film_2023_2024 = pd.concat(
    [df_2023, df_2024],
    ignore_index=True
)
```

```python
# Menampilkan data gabungan dari film tahun 2023 dan 2024
film_2023_2024
```

```python
# Melihat informasi dataset
film_2023_2024.info()
```

# Menganalisis 'Jumlah Film' per tahun

```python
# Menambahkan kolom Tahun dari Tanggal Tayang
film_2023_2024["Tahun"] = film_2023_2024["Tanggal Tayang"].str[-4:]
```

```python
# Menghitung jumlah film berdasarkan tahun
jumlah_film = film_2023_2024["Tahun"].value_counts().sort_index()

jumlah_film
```

```python
import matplotlib.pyplot as plt

# Membuat grafik jumlah film per tahun
jumlah_film.plot(kind="bar")

plt.title("Jumlah Film Indonesia 2023 vs 2024")
plt.xlabel("Tahun")
plt.ylabel("Jumlah Film")
plt.xticks(rotation=0)
plt.show()
```

# Menganalisis 'Rata-rata Penonton' per tahun

```python
# Mengubah Penonton menjadi angka
film_2023_2024["Penonton"] = (
    film_2023_2024["Penonton"]
    .astype(str)
    .str.replace(".", "", regex=False)
    .astype(int)
)
```

```python
# Melihat informasi dataset
film_2023_2024.info()
```

```python
# Menghitung rata-rata jumlah penonton per tahun
rata_penonton = (
    film_2023_2024.groupby("Tahun")["Penonton"]
    .mean()
    .round(0)
)

rata_penonton
```

```python
# Visualisasi rata-rata penonton per tahun
rata_penonton.plot(kind="bar")

plt.title("Rata-rata Jumlah Penonton Film 2023 vs 2024")
plt.xlabel("Tahun")
plt.ylabel("Rata-rata Penonton")
plt.xticks(rotation=0)
plt.show()
```

# Menganalisis "Genre" yang paling banyak diproduksi

```python
# Memisahkan genre yang memiliki lebih dari satu kategori
genre = film_2023_2024["Genre"].str.split(", ").explode()

# Menghitung jumlah film berdasarkan genre
jumlah_genre = genre.value_counts()

jumlah_genre
```

```python
import matplotlib.pyplot as plt

# Membuat grafik jumlah film berdasarkan genre
jumlah_genre.plot(kind="bar")

plt.title("Genre Film Indonesia 2023–2024")
plt.xlabel("Genre")
plt.ylabel("Jumlah Film")
plt.xticks(rotation=45)
plt.show()
```

# Menganalisis 'Total Penonton' berdasarkan 'Genre'

```python
# Mengubah Penonton menjadi angka
film_2023_2024["Penonton"] = (
    film_2023_2024["Penonton"]
    .astype(str)
    .str.replace(".", "", regex=False)
    .astype(int)
)
```

```python
# Melihat informasi dataset
film_2023_2024.info()
```

```python
# Memisahkan genre yang memiliki lebih dari satu kategori
df_genre_penonton = film_2023_2024.assign(
    Genre=film_2023_2024["Genre"].str.split(", ")
).explode("Genre")

# Menghitung total penonton berdasarkan genre
penonton_genre = (
    df_genre_penonton.groupby("Genre")["Penonton"]
    .sum()
    .sort_values(ascending=False)
)

penonton_genre
```

```python
# Membuat grafik total penonton berdasarkan genre
penonton_genre.plot(kind="bar")

plt.title("Total Penonton Berdasarkan Genre Film 2023–2024")
plt.xlabel("Genre")
plt.ylabel("Total Penonton")
plt.xticks(rotation=45)
plt.show()
```

# Menganalisis 'Film' dengan 'Penonton' terbanyak

```python
# Menampilkan 10 film dengan jumlah penonton terbanyak
film_10_teratas = film_2023_2024.sort_values(
    by="Penonton",
    ascending=False
).head(10)

film_10_teratas
```

# Menganalisis jumlah penonton terbanyak pada masing-masing tahun

```python
# Mencari film dengan penonton terbanyak di setiap tahun
film_terbanyak_tahun = (
    film_2023_2024.loc[
        film_2023_2024.groupby("Tahun")["Penonton"].idxmax()
    ]
)

film_terbanyak_tahun
```

```python
# Membuat grafik berdasarkan judul film
film_terbanyak_tahun.plot(
    x="Judul",
    y="Penonton",
    kind="bar"
)

plt.title("Film dengan Jumlah Penonton Terbanyak per Tahun")
plt.xlabel("Film")
plt.ylabel("Jumlah Penonton")
plt.xticks(rotation=0)
plt.show()
```
